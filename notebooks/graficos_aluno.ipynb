```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../dados/dados_alunos.csv')
df
```

```python
df['idade'].plot.hist(
    bins=10,
    title='Histograma de Idade dos Alunos',
    figsize=(6, 3 )
)
plt.xlabel('Idade')
plt.ylabel('frequência')
plt.show()
```

```python
# Agrupando dados

df['genero'].value_counts()
```

```python
df['genero'].value_counts().plot.bar(
    title='Gráfico - Gênero dos Alunos',
    figsize=(6, 3 )
)
plt.xlabel('Gênero')
plt.ylabel('Quantidade')
plt.xticks(rotation=0)
plt.show()
```

```python
df['genero'].value_counts().plot.barh(
    title='Gráfico - Gênero dos Alunos',
    figsize=(6, 3 )
)
plt.xlabel('Quantidade')
plt.ylabel('Gênero')
plt.xticks(rotation=0)
plt.show()
```

```python
df['genero'].value_counts().plot.pie(
    colors=["#135a0d","#50050c"],
    figsize=(5, 5), 
    autopct='%1.1f%%',
    wedgeprops=dict(width=0.7)   
)

plt.title('Quantidade de Alunos por Gênero')
plt.show()
```
